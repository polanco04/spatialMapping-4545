## 2. Writing and Reading Text Files

```python
# Writing to a text file
with open("example.txt", "w") as f:
    f.write("Hello, Jupyter!\n")
    f.write("This is a text file.\n")
```

```python
# Reading from a text file
with open("example.txt", "r") as f:
    contents = f.read()
contents
```

## 3. JSON Files

```python
import json

data = {"name": "Graylian", "type": "mascot", "skills": ["howl", "parrot talk"]}

# Write JSON
with open("data.json", "w") as f:
    json.dump(data, f)

# Read JSON
with open("data.json", "r") as f:
    loaded = json.load(f)
loaded
```

## 4. CSV Files

```python
import csv

# Writing CSV
rows = [["Name", "Score"], ["Alice", 90], ["Bob", 85]]
with open("scores.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(rows)

# Reading CSV
with open("scores.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

```python
import csv # same with json, not needed if cell above is ran beforehand

myRows = [["Name", "Hobby", "Favorite Movie"],
          ["Pedro", "Sports", "Infinity War"],
          ["Juan", "Reading", "Infinity Castle"]]

with open("people.csv", "w", newline="") as f:
  write = csv.writer(f)
  write.writerows(myRows)

with open("people.csv", "r") as f:
  read = csv.reader(f)
  for row in read:
    print(row)
```
