# 🧮 Sorting & Filtering in Pandas

## 🔹 LEARNING GOALS:
- Sort rows using `.sort_values()`
- Filter data using conditional logic
- Chain multiple filters together
- Use `.query()` for readable filtering


### 📥 1. Load the Dataset

```python
import pandas as pd

df = pd.read_csv("sample_data/students.csv")
df.head()
```

### 🔢 2. Sorting Data

```python
# Sort by math_score
df.sort_values(by="math_score")
```

```python
# Sort by science_score descending
df.sort_values(by="science_score", ascending=False)
```

```python
# Sort by multiple columns
df.sort_values(by=["science_score", "math_score"], ascending=[False, True])
```

### 🧪 3. Filtering with Conditions

```python
# Who scored at least 90 in math?
df[df["math_score"] >= 90]
```

```python
# Who scored less than 70 in either subject?
df[(df["math_score"] < 70) | (df["science_score"] < 70)]
```

### 🔗 4. Combining Filters

```python
# Students with B+ or higher in both
df[(df["math_score"] >= 87) & (df["science_score"] >= 87)]
```

### 🧾 5. Using `.query()` for Readability

```python
# Same query using .query()
df.query("math_score >= 87 and science_score >= 87")
```

### 🧠 Mini-Challenge

> Load `"data/survey.csv"` and:
- Sort by the most recent signup date
- Filter out anyone who didn't complete the survey (nulls or "N/A")
- Use `.query()` to select rows with a specific condition of your choice


```python
surveyData = {
    'name': ['Alice Johnson', 'Bob Smith', 'Charlie Brown', 'Diana Prince', 'Eve Davis',
             'Frank Miller', 'Grace Lee', 'Henry Wilson', 'Iris Chen', 'Jack Taylor'],

    'email': ['alice.j@email.com', 'bob.smith@email.com', 'charlie.b@email.com',
              'diana.p@email.com', 'eve.d@email.com', 'frank.m@email.com',
              'grace.l@email.com', 'henry.w@email.com', 'iris.c@email.com',
              'jack.t@email.com'],

    'signup_date': ['2024-01-15', '2024-01-18', '2024-01-20', '2024-02-03',
                    '2024-02-10', '2024-02-15', '2024-03-01', '2024-03-05',
                    '2024-03-12', '2024-03-20'],

    'survey_completed': [True, None, True, None, 'N/A', 'N/A', True, None, True, 'N/A']
}

surveyFrame = pd.DataFrame(surveyData)
surveyFrame
```

```python
surveyFrame.sort_values(by="signup_date", ascending=False)
```

```python
surveyFrame[(surveyFrame["survey_completed"].notna()) & (surveyFrame["survey_completed"] != 'N/A')]
```

```python
surveyFrame.query("survey_completed == True")
```

### 📝 Summary

| Task              | Code Example                                 |
|-------------------|-----------------------------------------------|
| Sort by column     | `df.sort_values(by="col")`                  |
